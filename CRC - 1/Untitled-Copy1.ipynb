```python
import pandas as pd

df = pd.read_csv("/home/Data_Drive_8TB/kykim/colon/TCR/GSE108989_CRC.bulk.exome.mutation.tab", sep='\t')
```

```python
msi_patients = ['P0413T', 'P0825T', 'P0123T', 'P0909T']
mss_patients = ['P0215T', 'P0411T', 'P0701T', 'P1012T', 'P1207T', 'P1212T', 'P1228T', 'P0309T']

def classify_cancer_type(patient):
    if patient in msi_patients:
        return 'MSI'
    elif patient in mss_patients:
        return 'MSS'

df['cancer_type'] = df['patient'].apply(classify_cancer_type)
```
